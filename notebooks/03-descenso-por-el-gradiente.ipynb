# **03. Descenso por el gradiente**

## **1. Descenso por el gradiente: encontrar el mínimo**

El descenso por el gradiente es el mecanismo fundamental que permite a una neurona, y a una red neuronal entera, aprender. La idea es sencilla: queremos encontrar el mínimo de una función. Para ello, nos fijamos en la pendiente de la función en un punto y nos movemos en la dirección contraria a la pendiente. Repetimos este proceso muchas veces hasta que llegamos a un mínimo.

Tomemos una función polinómica sencilla, como por ejemplo:

$$f(x)=x^2-2x+2$$

Su mínimo está en x=1 y vale 1. Podemos comprobarlo calcularlo fácilmente así:

$$f'(x)=2x-2=0 \Rightarrow x=1 \Rightarrow f(1)=1$$

Sin embargo, a veces las funciones en las que tenemos que buscar un mínimo no son tan sencillas. Por ejemplo, si tuviéramos que encontrar el mínimo de una función complicada, no podríamos resolver la ecuación $f'(x)=0$ de forma directa. En esos casos, necesitamos un método que nos permita aproximarnos al mínimo paso a paso.

Una primera estrategia sería partir de un punto elegido al azar y mover x en pasos fijos pequeños. Supongamos que empezamos en x=2.5. Nos moveríamos un poco hacia la izquierda o hacia la derecha, dependiendo de la pendiente de la función en ese punto. Si la pendiente es negativa, significa que estamos por encima del mínimo y debemos movernos hacia la izquierda; si es positiva, debemos movernos hacia la derecha.

Por tanto, desde 2.5 podemos ir desplazándonos en dirección opuesta al gradiente en pequeños pasos, por ejemplo 0.01 e ir comprobando si la función disminuye. Mientras siga disminuyendo, seguimos en esa dirección. Si en algún momento la función aumenta, significa que hemos alcanzado el mínimo.

```python
def f(x):
    return x**2 - 2*x + 2

x = 2.5
paso = 0.01
contador = 0

while f(x - paso) < f(x) and contador < 1000:
    x -= paso
    contador += 1

print("Aproximación:", x)
print("Pasos:", contador)
```

Hay un método mejor que dar siempre el mismo paso: la **pendiente**. A la derecha del mínimo es positiva; a la izquierda es negativa. Justo en el mínimo es cero. También podemos ver en el ejemplo que la pendiente es mayor cuanto más lejos estamos del mínimo. Si utilizamos este hecho para movernos, podemos hacer pasos más grandes cuando estamos lejos del mínimo y pasos más pequeños cuando nos acercamos. Esto nos permite llegar al mínimo de manera más eficiente. Pero habíamos dicho que calcular la pendiente de una función complicada puede ser difícil. Por eso vamos a ver cómo aproximarla mediante una pequeña diferencia.

$$f'(x)\approx\frac{f(x+\delta)-f(x-\delta)}{2\delta}$$

Calculamos la función un poco a la derecha y un poco a la izquierda y dividimos por la separación. Es la idea de «cambio vertical dividido por cambio horizontal» que usamos al interpretar una pendiente.

```python
def f(x):
    return x**2 - 2*x + 2

x = 2.5
delta = 0.01
gamma = 0.45

for paso in range(7):
    pendiente = (f(x + delta) - f(x - delta))/(2 * delta)
    x = x - gamma * pendiente
    
    print("Paso", paso + 1, "x:", x, "f(x):", f(x))
```

La instrucción central es `x = x - gamma * pendiente`. Usa el x actual para calcular el nuevo x. El símbolo que utilizaremos en las fórmulas, $\leftarrow$, significa precisamente «actualiza este valor»:

$$x\leftarrow x-\gamma f'(x).$$

Después de siete pasos ya estamos muy cerca de 1. $\gamma = 0.45$ («gamma») es una constante que nos permite ajustar la rapidez con la que nos movemos. Si es demasiado grande, podemos saltarnos el mínimo; si es demasiado pequeña, tardaremos mucho en llegar.

## **4. La tasa de aprendizaje**

$\gamma$ (se lee «gamma») controla el tamaño del ajuste y se llama **tasa de aprendizaje**. Es un valor que elegimos. Una tasa pequeña puede necesitar muchos pasos. Una tasa grande puede hacernos saltar de un lado al otro del valle o alejarnos cada vez más.

Mira la siguiente figura. Parecen dos curvas iguales, pero si te fijas en la escala del eje horizontal, verás que no lo son. La derivada en un punto concreto (x=2 en la curva izquierda y x=0.2 en la curva derecha) es la misma en ambas curvas, pero el efecto de un desplazamiento de una cierta cantidad basado en la derivada no será el mismo en ambas curvas. Por eso, la tasa de aprendizaje es útil. Nos permite ajustar el tamaño del paso que damos en cada iteración, y así controlar cómo nos acercamos al mínimo.

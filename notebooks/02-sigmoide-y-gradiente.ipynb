# **02. Neurona con activación sigmoide**

## **1. Una nueva función de activación**

La función de activación que hemos estado utilizando hasta ahora es la función escalón $x \geq 0$, que devuelve 1 si la suma ponderada de las entradas es mayor o igual a un umbral $\theta$, y 0 en caso contrario. Esto está representado por la siguiente ecuación:

$$\text{Salida} = \begin{cases}
1 & \text{si } \Sigma_{i=1}^n w_i e_i \geq \theta\\
0 & \text{en caso contrario}
\end{cases}$$

Realmente, la función escalón no devuelve un 1 o un 0, sino que devuelve un valor booleano (verdadero o falso) que nosotros interpretamos como 1 o 0.

Mantener esta función nos ha permitido construir un modelo de neurona que puede clasificar datos, pero tiene algunas limitaciones. Por ejemplo, no nos permitirá calcular un gradiente para ajustar los pesos de la neurona durante el entrenamiento, que veremos más adelante.

Por tanto, vamos a hacer un primer cambio a nuestra neurona. Vamos a mover el umbral $\theta$ al otro lado de la desigualdad.

$$
\sum_{i=1}^n w_i e_i - \theta \geq 0
$$

Ahora renombraremos a $-\theta$ como $w_0$. Esto nos permite escribir la condición de activación de la neurona como una suma ponderada de entradas y pesos, incluyendo el umbral como un peso más.

$$
\sum_{i=1}^n w_i e_i + w_0 \geq 0
$$

Si consideramos que $w_0$ está multiplicado por una entrada $e_0$ que siempre vale 1, podemos escribir la condición de activación como una suma ponderada de entradas y pesos, incluyendo el umbral como un peso más. Fíjate que ahora la suma empieza en $i=0$ y que $w_0$ es el peso correspondiente a la entrada $e_0$.

$$
\sum_{i=0}^n w_i e_i \geq 0
$$

Ten en cuenta que esta nueva expresión es totalmente equivalente a la anterior.

## **2. Función sigmoide**

Lo siguiente que vamos a hacer en nuestra neurona es cambiar la función de activación $x \geq 0$ por una función continua y derivable. La función que vamos a usar es la **sigmoide**. La sigmoide es una función que nos da un valor entre 0 y 1, y que cambia suavemente de 0 a 1 a medida que su entrada pasa de valores negativos a positivos. Esto nos permite tener una transición suave en lugar de un salto brusco. La expresión de la función sigmoide es:

$$\sigma(x)=\frac{1}{1+e^{-x}}.$$

¡OJo! aquí $e$ es la constante matemática aproximadamente igual a $2.718\text{...}$, no una entrada. La letra griega $\sigma$ se lee «sigma». La siguiente figura muestra la forma de la función sigmoide. Fíjate en que para valores negativos de $x$, la salida es cercana a 0, y para valores positivos de $x$, la salida es cercana a 1. En el punto $x=0$, la salida es exactamente 0.5.

<img src="imgs/sigmoide.png" alt="La sigmoide transforma cualquier número real en un valor entre 0 y 1." width="650" style="max-width:100%;height:auto;display:block;margin-left:auto;margin-right:auto;">

Recuerda que la función escalón $x \ge 0$ que hemos usado hasta ahora cambia bruscamente de 0 a 1 en cero, mientras que la función sigmoide cambia suavemente de 0 a 1 en el entorno de cero. Realmente, la función escalón no devuelve un $0$, ni un $1$, sino un valor booleano verdadero o falso. Somos nosotros los que interpretamos el 0 como falso y el 1 como verdadero.

<img src="imgs/mayorigual.png" alt="El escalón cambia bruscamente de 0 a 1 en cero." width="650" style="max-width:100%;height:auto;display:block;margin-left:auto;margin-right:auto;">

Veamos un ejemplo de cómo se comporta la función sigmoide. En el siguiente código, vamos a calcular la salida de la función sigmoide para una serie de valores de entrada.

```python
from math import exp

def sigmoide(x):
    return 1 / (1 + exp(-x))

for x in [-10, -1, -0.1, 0, 0.1, 1, 10]:
    print("x =", x, "→ salida =", round(sigmoide(x), 4))
```

### **Una derivada que resulta cómoda**

La sigmoide tiene una derivada muy cómoda de calcular, ya que su derivada se puede expresar en términos de la propia función sigmoide:

$$\sigma'(x)=\sigma(x)\,[1-\sigma(x)]$$

Si ya conocemos la salida $o=\sigma(x)$, basta calcular $o(1-o)$. Ya veremos la utilidad de esta derivada cuando hablemos del entrenamiento de la neurona. Por ahora, nos basta con saber que la sigmoide es derivable en todo su dominio, y que su derivada es fácil de calcular.

## **5. Modelo de la neurona con activación sigmoide**

Volvamos a nuestra neurona. Teníamos la comparación:

$$ \sum_{i=0}^{2}w_i e_i \geq 0$$

Ahora, en lugar de usar la función escalón, vamos a usar la función sigmoide para obtener una salida suave. La salida de la neurona será:

$$\text{Salida} = \sigma(w_0e_0+w_1e_1+w_2e_2) = \sigma\left(\sum_{i=0}^{2}w_i e_i\right)$$

Recuerda que la función sigmoide es:

$$\sigma(x)=\frac{1}{1+e^{-x}}$$

<img src="imgs/perceptron.svg" alt="La entrada constante e₀ = 1 permite incluir el sesgo como w₀." width="650" style="max-width:100%;height:auto;display:block;margin-left:auto;margin-right:auto;">

En el siguiente código podemos ver cómo se comporta la neurona con activación sigmoide. Vamos a calcular la salida de la neurona para una serie de valores de entrada y pesos, y observar cómo cambia la salida a medida que variamos los pesos.

```python
from math import exp

def sigmoide(x):
    return 1 / (1 + exp(-x))

# Neurona con activación escalón 

def neurona_escalon(e1, e2):
    w1 = -1/2
    w2 = 1
    umbral = 1
    suma = w1 * e1 + w2 * e2
    if suma >= umbral:
        salida = 1
    else:
        salida = 0
    return int(salida)

# Neurona con activación sigmoide

def neurona_sigmoide(e1, e2):
    w0 = -1
    w1 = -1/2
    w2 = 1
    z = w0 + w1 * e1 + w2 * e2
    return sigmoide(z)


puntos = {"a": (2, 2), "b": (1, 3), "c": (2, 4),
          "d": (3, 1), "e": (5, 2)}

for nombre, (e1, e2) in puntos.items():
    salida_escalon = neurona_escalon(e1, e2)
    salida_sigmoide = neurona_sigmoide(e1, e2)
    print(f"Punto {nombre}: entrada=({e1}, {e2}), salida escalón={salida_escalon}, salida sigmoide={salida_sigmoide:.4f}")
```

Ahora, nuestra neurona con activación sigmoide nos devuelve un valor entre 0 y 1, que podemos interpretar como una puntuación que si es mayor o igual a 0.5, podemos considerar que la neurona predice la clase 1, y si es menor que 0.5, predice la clase 0.

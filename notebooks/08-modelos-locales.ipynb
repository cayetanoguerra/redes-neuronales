## **1. ¿Dónde está el modelo?**

Cuando escribimos en ChatGPT, nuestro mensaje viaja por internet hasta un centro de datos. Allí, un modelo con una cantidad enorme de parámetros calcula la respuesta utilizando ordenadores especializados, y el resultado vuelve a nuestra pantalla. Decimos que el modelo se ejecuta **en la nube**: lo utilizamos, pero no lo tenemos.

Un **modelo local** se ejecuta en el mismo ordenador en el que escribimos. Para ello necesitamos dos cosas:

1. **El modelo**, que en la práctica es un archivo (a veces varios) con los parámetros aprendidos durante el entrenamiento. Recuerda que un parámetro es un número que interviene en los cálculos de la red neuronal. Un modelo con 4 000 millones de parámetros es, esencialmente, un archivo con 4 000 millones de números.
2. **Un programa que sepa usarlo**, es decir, que lea esos números y realice las operaciones de la red para generar texto. A esta fase de uso la llamamos **inferencia**.

LM Studio reúne las dos piezas: ayuda a conseguir el archivo del modelo y contiene el programa que realiza la inferencia.

¿De dónde salen esos archivos? Varias empresas y grupos de investigación publican los parámetros de algunos de sus modelos para que cualquiera pueda descargarlos. Se les suele llamar **modelos de pesos abiertos** (*open-weight*). Entre las familias más conocidas están Gemma (Google), Llama (Meta), Qwen (Alibaba), Mistral (Mistral AI), Phi (Microsoft) o gpt-oss (OpenAI). Cada modelo se publica con una licencia que indica cómo se puede utilizar; conviene leerla si vamos a usarlo fuera del ámbito personal.

<b>Un modelo local es un modelo de lenguaje cuyos parámetros están guardados en tu ordenador y que se ejecuta en él, sin necesidad de enviar tus textos a un servidor externo.</b>

## **2. ¿Por qué usar un modelo local?**

Si ya tenemos ChatGPT, ¿para qué complicarnos? Hay buenas razones, especialmente en un centro educativo:

- **Privacidad.** El texto que escribimos no sale del ordenador. Esto es importante cuando trabajamos con información de nuestro alumnado: calificaciones, respuestas de un examen, informes o cualquier dato personal. Con un modelo local, esa información no se comparte con terceros.
- **Funciona sin conexión.** Una vez descargado, el modelo responde aunque no haya internet. Útil en un aula con mala conexión o en una salida.
- **Sin coste por uso.** No hay suscripción ni se paga por cada consulta. Podemos lanzar cientos de peticiones, por ejemplo para corregir un examen completo, sin preocuparnos por la factura. El único coste es el del ordenador y la electricidad.
- **Control.** Elegimos qué modelo y qué versión usar, y sigue siendo el mismo mañana. Un servicio en la nube puede cambiar el modelo sin avisar.
- **Aprender cómo funciona.** Ver el tamaño del archivo, cuánta memoria ocupa o cuántos tokens por segundo genera hace tangibles las ideas del notebook anterior.

Pero también tienen limitaciones que conviene conocer:

- **Son más pequeños.** Un portátil no puede alojar modelos tan grandes como los de los centros de datos. Los modelos locales habituales tienen entre 1 000 y 30 000 millones de parámetros. Responden bien a muchas tareas, pero razonan peor, saben menos cosas y **se equivocan con más frecuencia**.
- **Dependen del hardware.** Necesitan memoria suficiente y, para ir rápido, una buena tarjeta gráfica o un procesador como los Apple Silicon. En un ordenador modesto, las respuestas pueden tardar.
- **Su conocimiento está congelado.** Saben lo que aprendieron hasta la fecha de su entrenamiento y, por sí solos, no buscan en internet.
- **Requieren algo de mantenimiento.** Elegimos, descargamos y actualizamos los modelos nosotros mismos.

| | Modelo en la nube (p. ej. ChatGPT) | Modelo local (p. ej. con LM Studio) |
|---|---|---|
| Dónde se ejecuta | Servidores de una empresa | Tu ordenador |
| ¿Tus textos salen del equipo? | Sí | No |
| Conexión a internet | Imprescindible | Solo para descargar el modelo |
| Coste | Suscripción o pago por uso | Gratuito tras la descarga |
| Tamaño y capacidad | Modelos muy grandes, más capaces | Modelos pequeños o medianos |
| Velocidad | Depende del servicio | Depende de tu ordenador |

## **3. ¿Cuánto ocupa un modelo?**

Ya sabemos que un modelo es, sobre todo, una gran colección de números. Para saber cuánto ocupa basta con multiplicar dos cantidades: **cuántos parámetros tiene** y **cuánto espacio ocupa cada uno**.

Durante el entrenamiento, cada parámetro se suele guardar con 16 bits, es decir, 2 bytes. Un modelo de 4 000 millones de parámetros (lo que se escribe abreviado como **4B**, de *billion*, «mil millones» en inglés) ocuparía entonces:

$$4\,000\,000\,000 \times 2 \text{ bytes} = 8\,000\,000\,000 \text{ bytes} \approx 8 \text{ GB}$$

Y el modelo tiene que caber entero en la memoria del ordenador mientras lo usamos. Para un portátil corriente, 8 GB solo para el modelo es mucho.

La solución se llama **cuantización**: guardar cada parámetro con menos bits, por ejemplo 4 en lugar de 16. Es parecido a redondear: en lugar de guardar 0,73418 guardamos un valor aproximado que se puede escribir con menos cifras. Se pierde algo de precisión, pero el modelo sigue funcionando sorprendentemente bien y ocupa unas cuatro veces menos. Con 4 bits (medio byte) por parámetro, nuestro modelo de 4B pasaría a ocupar unos 2 GB.

<div style="display:inline-block; background-color:#fff3cd; border:1px solid #f0c36d; border-left:6px solid #e0a800; border-radius:6px; padding:12px 16px; margin:12px 0; color:#333;">
<b>Tamaño aproximado de un modelo = número de parámetros × bits por parámetro ÷ 8. La cuantización reduce los bits por parámetro para que el modelo quepa en la memoria de un ordenador normal.</b>
</div>

Puedes comprobar el cálculo con la siguiente celda. Cambia el número de parámetros o los bits y observa cómo varía el tamaño.

```python
def tamano_modelo_gb(parametros, bits_por_parametro):
    """Tamaño aproximado en GB de los parámetros de un modelo."""
    bytes_totales = parametros * bits_por_parametro / 8
    return bytes_totales / 1e9

for nombre, parametros in [("Modelo de 1B", 1e9), ("Modelo de 4B", 4e9),
                           ("Modelo de 8B", 8e9), ("Modelo de 27B", 27e9)]:
    print(f"{nombre:>14}:  16 bits → {tamano_modelo_gb(parametros, 16):5.1f} GB"
          f"   |   4 bits → {tamano_modelo_gb(parametros, 4):5.1f} GB")
```

Los archivos reales ocupan algo más que este cálculo, porque algunas partes del modelo se guardan con más precisión y porque incluyen información adicional. Además, mientras el modelo genera texto necesita memoria extra para el contexto de la conversación. Como orientación:

| Memoria del ordenador (RAM) | Modelos que podemos usar con comodidad (cuantizados a 4 bits) |
|---|---|
| 8 GB | Hasta unos 4B parámetros |
| 16 GB | Hasta unos 8B–12B parámetros |
| 32 GB o más | Modelos de 20B–30B parámetros |

En los Mac con Apple Silicon la memoria es compartida entre el procesador y la parte gráfica, así que cuenta toda la RAM. En un PC con tarjeta gráfica, el modelo irá mucho más rápido si cabe en la memoria de la tarjeta (VRAM).

## **4. Cómo leer el nombre de un modelo**

Los nombres de los modelos parecen jeroglíficos, pero casi siempre siguen el mismo patrón. Tomemos dos ejemplos que aparecerán en las capturas: `Qwen3 4B 2507 · 4bit · MLX` y `Gemma 4 E4B · Q4_K_M · GGUF`.

| Parte del nombre | Qué significa |
|---|---|
| **Qwen3**, **Gemma 4** | Familia y versión del modelo. |
| **4B**, **12B**, **27B** | Número de parámetros en miles de millones (*billions*). |
| **E4B** | «4B efectivos»: el modelo tiene más parámetros en total, pero está diseñado para funcionar con un coste de cálculo y memoria parecido al de un modelo de 4B. |
| **2507** | Fecha de la versión (en este caso, julio de 2025). |
| **Instruct** | Ajustado para seguir instrucciones y conversar. Es lo que normalmente queremos. |
| **Thinking** | Versión que «piensa en voz alta» antes de contestar: escribe un razonamiento previo y luego la respuesta. Suele acertar más en problemas, pero tarda más. |
| **4bit**, **Q4_K_M**, **Q8_0** | Tipo de cuantización. El número indica aproximadamente los bits por parámetro: Q4 ocupa menos, Q8 es más fiel al original. |
| **GGUF** | Formato de archivo del programa llama.cpp. Funciona en Windows, Linux y Mac. |
| **MLX** | Formato optimizado para los Mac con Apple Silicon. En esos equipos suele ser más rápido. |

LM Studio también muestra unos iconos con las **capacidades** de cada modelo:

- 👁 **Vision**: puede recibir imágenes además de texto.
- 🔨 **Tool use**: puede pedir que se utilicen herramientas externas (por ejemplo, una calculadora o una búsqueda). Lo veremos en el bloque de programación de APIs.
- 💡 **Reasoning**: es capaz de razonar antes de responder.

Lee con atención la respuesta de la captura. Le hemos preguntado qué es un modelo de lenguaje **local** y ha contestado, con total seguridad, que es un modelo adaptado a un idioma o una región. ¡Ha interpretado «local» en el sentido de «de un lugar»!

#### El panel de configuración

A la derecha aparece el panel **Model Parameters**. Los ajustes más útiles son:

- **System Prompt**: instrucciones generales que el modelo recibe al principio de cada conversación, por ejemplo «Eres un profesor de Biología de 3.º de ESO. Responde siempre en español y con ejemplos sencillos». Forman parte del contexto, aunque no aparezcan en el chat.
- **Temperature** (temperatura): controla cuánto azar hay al elegir el siguiente token. Con valores bajos (0,1–0,3), el modelo elige casi siempre los tokens más probables y sus respuestas son más previsibles; con valores altos (0,8–1), es más variado y creativo, pero también más propenso a desvariar.
- **Top K** y **Top P Sampling**: limitan entre cuántos tokens candidatos se elige. Top K = 20 significa que solo se consideran los 20 tokens más probables.
- **Limit Response Length**: fija un máximo de tokens para la respuesta.

```python
import json
import urllib.request

direccion = "http://localhost:1234/v1/chat/completions"
modelo = "qwen/qwen3-4b-2507"   # Cámbialo por el "API Model Identifier" de tu modelo

peticion = {
    "model": modelo,
    "messages": [
        {"role": "system", "content": "Responde en español, en un máximo de dos frases."},
        {"role": "user", "content": "¿Qué ventaja tiene ejecutar un modelo de lenguaje en mi propio ordenador?"}
    ],
    "temperature": 0.3
}

try:
    solicitud = urllib.request.Request(
        direccion,
        data=json.dumps(peticion).encode("utf-8"),
        headers={"Content-Type": "application/json"}
    )
    with urllib.request.urlopen(solicitud, timeout=120) as respuesta:
        datos = json.load(respuesta)
    print(datos["choices"][0]["message"]["content"])
except Exception as error:
    print("No se ha podido conectar con LM Studio.")
    print("Comprueba que está abierto, que el servidor está en marcha y que hay un modelo cargado.")
    print("Detalle:", error)
```

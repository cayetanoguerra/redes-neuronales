**Fundamentos de la Inteligencia Artificial y Programación de APIs para LLMs**

# **07. Redes neuronales para texto: LLMs**


Imagina que puedes pedirle a un programa que te explique un concepto difícil, que te ayude a entender un documento o que prepare un ejemplo para estudiar. No necesitas aprender una lista de comandos: escribes lo que quieres con tus propias palabras y recibes una respuesta que puedes seguir comentando. Esa es la experiencia que ofrecen herramientas como ChatGPT.

¿Cómo consigue una máquina responder de esa manera? Para entenderlo vamos a empezar por algo que hacemos casi sin pensar: completar una frase. Desde ahí iremos descubriendo cómo se pueden aprender regularidades del lenguaje a partir de muchos ejemplos y utilizarlas para generar respuestas.

No necesitas saber programar ni recordar fórmulas complicadas. Nos apoyaremos en situaciones cotidianas y en algunas ideas matemáticas sencillas. Al terminar podrás explicar qué es un modelo de lenguaje, qué ocurre cuando le escribes y por qué a veces puede equivocarse aunque su respuesta parezca muy convincente.

## **1. Conversar con un programa**

El 30 de noviembre de 2022 OpenAI presentó ChatGPT al público. Muchas personas descubrieron entonces que podían conversar con un programa de una forma sorprendentemente natural. Bastaba con escribir una pregunta para recibir una explicación y, si algo no quedaba claro, pedir que lo contara de otra manera.

Podemos escribir, por ejemplo: «Explícame qué es una célula como si fuera la primera vez que lo estudio». Si la respuesta nos resulta demasiado técnica, añadimos: «Utiliza un ejemplo más cotidiano». Y, después, podemos pedir: «Ahora hazme una pregunta para comprobar si lo he entendido». Cada intervención ayuda a orientar lo que viene a continuación.

Esta forma de trabajar permite expresar objetivos que serían difíciles de recoger en un formulario. No siempre sabemos de antemano qué explicación necesitamos. A veces lo descubrimos al leer una primera respuesta y hacer una nueva pregunta. La conversación nos permite ir afinando la petición.

Cuando el resultado está bien escrito, es fácil sentir que al otro lado hay alguien que entiende lo que queremos y sabe del tema. Esa impresión resulta comprensible: estamos acostumbrados a relacionar el lenguaje con personas. Para utilizar la herramienta con criterio, conviene conocer qué hay detrás de esa facilidad para conversar.

Vamos a seguir el camino de una pregunta desde que la escribimos hasta que aparece la respuesta. Por el camino descubriremos que una tarea aparentemente sencilla, anticipar qué viene después en un texto, puede dar lugar a capacidades mucho más amplias de lo que parece.

## **2. La aplicación y su motor**

ChatGPT permite trabajar con lenguaje: hacer preguntas, redactar, resumir o transformar un texto. Para producir sus respuestas utiliza modelos (grandes redes neuronales) que han aprendido regularidades a partir de grandes cantidades de datos. Gracias a ese aprendizaje pueden construir explicaciones que se parecen a las que escribiría una persona.

Ahora bien, una explicación clara también puede contener errores. Si preguntamos por una fecha, un autor o un resultado científico, el tono seguro de la respuesta no basta para saber si es correcto. Más adelante entenderemos por qué puede ocurrir y cómo influye en la forma de utilizar estas herramientas.

Para empezar, distingamos la aplicación que utilizamos del modelo que produce el lenguaje. La interfaz de ChatGPT es la parte con la que interactuamos: ahí escribimos los mensajes y leemos las respuestas. El modelo (red neuronal) es un componente del sistema que realiza los cálculos necesarios para generar texto.

Podemos compararlo con una aplicación de mapas. En la pantalla vemos un mapa, elegimos un destino y recibimos una ruta. Detrás hay otros componentes que procesan esa petición. Algo parecido sucede aquí: la ventana de conversación facilita el intercambio, mientras que el modelo realiza una parte esencial del trabajo que lo hace posible. Más adelante veremos que el modelo puede ser utilizado en otras aplicaciones, no solo en la interfaz de ChatGPT. Lo veremos cuando hablemos de la API de OpenAI.

Además, una aplicación puede incorporar herramientas para buscar información o realizar cálculos. Conviene recordar esta distinción: las capacidades de un sistema completo dependen tanto del modelo como de los componentes con los que se conecta.

En este recorrido nos centraremos en los modelos llamados **GPT**. Sus siglas corresponden a *Generative Pretrained Transformer*, que podemos traducir como «transformer generativo preentrenado». Por ahora puede parecer un nombre complicado. Lo iremos descomponiendo hasta que cada palabra tenga sentido.

## **3. Qué es un modelo de lenguaje**

LLM significa Large Language Model, es decir, «gran modelo de lenguaje». GPT pertenece a esa familia. Para comprender el nombre, empecemos por la palabra «modelo».

Un modelo es una representación que nos permite estudiar algo o calcular un resultado. Un modelo meteorológico utiliza información sobre la atmósfera para estimar cómo puede evolucionar el tiempo. Un modelo de lenguaje utiliza el contexto disponible para estimar qué continuaciones son plausibles en un texto.

<b>Un modelo del lenguaje es un predictor de palabras que, a partir de lo que ya se ha escrito, asigna probabilidades a las posibles siguientes palabras.</b>

Piensa en la frase «El gato se ha subido al…». Algunas continuaciones resultan más esperables que otras. «Tejado» encaja en muchas situaciones; «jueves» sería bastante más extraña. Un modelo de lenguaje aprende relaciones que le permiten asignar probabilidades diferentes a esas posibilidades.

Los modelos que nos interesan realizan ese cálculo mediante redes neuronales. Una red neuronal contiene parámetros: valores numéricos que se ajustan durante el aprendizaje y determinan cómo se transforma la entrada en un resultado. El término «grande» hace referencia a su escala, especialmente a la cantidad de parámetros y a los recursos y datos utilizados en su desarrollo.

Cuando escribimos «¿Capital de Francia?», podemos obtener «París». Desde fuera solo vemos una pregunta y una respuesta. La misma herramienta puede recibir una petición más abierta, como «Explica qué diferencia hay entre una ciudad y un país para un niño de ocho años».

El modelo puede responder utilizando lo aprendido durante su entrenamiento. Eso no significa que consulte una ficha o una página web cada vez que contesta. Si interviene una búsqueda externa, se añade otro mecanismo. Esta diferencia cobra especial importancia cuando necesitamos información reciente o queremos comprobar de dónde procede una afirmación.

## **7. Completar una frase**

Si abres, por ejemplo, el WhatsApp, seguramente verás que el teclado propone palabras mientras escribes. Si empiezas con «Buenos», puede sugerirte «días». Si escribes «Muchas», quizá aparezca «gracias». Las sugerencias aprovechan una propiedad del lenguaje: unas combinaciones son mucho más habituales que otras. Te propongo un pequeño experimento: selecciona al azar cualquiera de las palabras que te sugiere el teclado y continúa escribiendo. Verás que, a medida que avanzas, las sugerencias se van ajustando a lo que ya has escrito. El texto del ejemplo que ves en la figura se ha hecho así. Si te fijas, gramaticalmente es correcto, pero semánticamente no tiene mucho sentido. El Whatsapp tiene un predictor de palabras, pero no es de generar frases con sentido por su cuenta. Aun así, nos ofrece un buen punto de partida para comprender lo que es un modelo del lenguaje.

Un gran modelo de lenguaje utiliza representaciones mucho más complejas y puede trabajar con un contexto mucho más amplio que un predictor sencillo.

Prueba a completar esta frase antes de seguir leyendo: «Tengo macarrones en la nevera, así que hoy me quedaré a comer en…».

Es bastante probable que hayas pensado en «casa». ¿Por qué? La palabra «en» permite muchísimas continuaciones. Sin embargo, saber que hay comida en la nevera y que alguien va a quedarse a comer hace que «casa» encaje especialmente bien.Ahora imagina que antes te han contado que en la oficina hay una cocina y que han dejado allí la comida. Con esa información, «la oficina» también sería una continuación razonable. Las mismas palabras finales pueden conducir a respuestas distintas cuando cambia el contexto. No estamos buscando una única palabra permitida, sino valorando qué alternativas resultan más plausibles con la información disponible. Esa es una idea central para entender un modelo de lenguaje. Nosotros realizamos este ejercicio apoyándonos en nuestra experiencia del mundo. El modelo aprende asociaciones a partir de datos y las utiliza mediante cálculos numéricos. Vamos a ver cómo se puede empezar a construir esa capacidad a partir de algo muy sencillo: contar palabras.

## **8. Las palabras más frecuentes**

Si reunimos una colección de textos y contamos sus palabras, encontraremos muchas veces «de», «la», «que» o «en». Estas palabras participan en una enorme variedad de frases. Otras, como «microscopio», aparecerán con mucha menos frecuencia en una colección general. Con ese recuento podemos calcular la frecuencia de cada palabra. Si una aparece cien veces en un conjunto de mil palabras, representa el diez por ciento del total. Ya tenemos una primera descripción estadística del lenguaje de esos textos. ¿Serviría para escribir una frase? Imagina que siempre elegimos la palabra más frecuente. El resultado sería repetitivo y no tendría sentido. Que «de» aparezca muchas veces no la convierte en una buena continuación de «me quedaré a comer en». Además, las frecuencias dependen del material que hayamos reunido. En una colección de informes médicos encontraremos términos distintos de los habituales en las noticias deportivas. Para interpretar un recuento necesitamos saber de qué textos procede y cómo se ha realizado. Incluso hay decisiones que parecen pequeñas y cambian el resultado. Podemos contar «tengo» y «tener» por separado o agrupar ambas formas bajo el verbo «tener». Según lo que queramos estudiar, una opción será más útil que otra.

Contar palabras es un comienzo, pero nos falta la relación entre ellas. Para continuar una frase necesitamos saber qué suele aparecer después de lo que ya está escrito. Vamos a pasar de observar palabras aisladas a observar sus combinaciones.

## **9. Reunir ejemplos en un corpus**

Una colección de textos preparada para su estudio o procesamiento recibe el nombre de corpus. Puede contener noticias, libros, conversaciones o documentos de otros tipos. Es el conjunto de ejemplos sobre el que podemos observar cómo se utiliza el lenguaje. Tomemos una noticia sobre un torneo de tenis. Podemos leerla para informarnos de los jugadores y las fechas. También podemos tratarla como una secuencia de palabras: dónde empieza una frase, qué expresiones se repiten y qué palabras aparecen juntas. Al reunir muchos textos, encontramos patrones que un único artículo apenas permitiría observar. La elección de los ejemplos influye en el aprendizaje. Si abundan los textos de un estilo, el modelo tendrá muchas oportunidades de aprender sus regularidades. Si una variedad de lengua o un tema están poco representados, tendrá menos ejemplos sobre los que apoyarse También importa cómo preparamos el material. Un corpus puede contener duplicados, fragmentos incompletos o texto deteriorado. Limpiarlo y organizarlo cambia la información que recibe el modelo. Acumular más datos no hace desaparecer automáticamente esos problemas. Por ahora utilizaremos palabras completas para que los ejemplos sean fáciles de seguir. Más adelante veremos que los modelos GPT trabajan con unidades llamadas tokens, que pueden ser palabras o partes de ellas. La idea principal se mantiene: tenemos secuencias y queremos aprender regularidades que nos ayuden a continuarlas.

## **10. Qué suele venir después de una palabra**

Leamos este comienzo: «Moderno, tecnológico e innovador». Si miramos las palabras de dos en dos obtenemos «moderno tecnológico», «tecnológico e» y «e innovador». Si seguimos con «el lema es nítido», aparecen «el lema», «lema es» y «es nítido». Cada par de unidades consecutivas se llama bigrama. A partir de muchos textos podemos contar qué bigramas se repiten y utilizar esa información para estimar qué palabra suele seguir a otra.

<div style="text-align:center">

</div>

Por ejemplo, después de «tengo» pueden aparecer «que», «más» o «catarro». Si conocemos cuántas veces sucede cada cosa en nuestro corpus, podemos asignar probabilidades diferentes a esas continuaciones.

Imaginemos un corpus pequeño, en el que tenemos un vocabulario de solo mil palabras. En ese caso, hay un millón de pares posibles de palabras consecutivas. Supongamos que hemos reunido un conjunto de textos en el que «tengo» aparece con las siguientes palabras: «que», «más» y «catarro». En sesenta ocasiones sigue «que», en veinticinco sigue «más» y en quince sigue «catarro». En este ejemplo inventado, la probabilidad estimada de «que» después de «tengo» sería 0.00034, de la misma manera que la de «más» sería 0.000012 y la de «catarro» 0.0000045. Observa que habrá combinaciones que no se den nunca en nuestro corpus.

Ya tenemos una forma elemental de anticipar palabras. Su principal limitación es que solo utiliza una palabra anterior. Para mejorar la predicción necesitaremos mirar un poco más atrás.

## **11. Más contexto y muchas más posibilidades**

Después de «tengo» puede venir «que». Pero, una vez que hemos escrito «tengo que», las alternativas cambian. Ahora encajan «hacer», «ver» o «ir». Si solo miráramos la última palabra, «que», perderíamos parte de la información que ayuda a elegir la continuación. Podemos contar grupos de tres unidades, llamados trigramas. Así estudiamos combinaciones como «tengo que hacer» o «tengo que ir». En términos de predicción, utilizamos las dos palabras anteriores para estimar la siguiente. En general, disponer de más contexto ayuda a distinguir situaciones. «El banco del parque» y «el banco concedió» contienen la misma palabra, pero nos llevan a significados distintos. Lo que aparece alrededor permite interpretar mejor lo que leemos y lo que podría venir después.

El problema es que el número de combinaciones crece muy deprisa. Con mil palabras, los pares sumaban un millón de posibilidades. Para las tripletas tenemos mil por mil por mil: mil millones. Si añadimos una cuarta posición, llegamos a un billón, es decir, un millón de millones. No solo necesitaríamos una tabla enorme. También nos faltarían ejemplos para llenar muchas de sus casillas. Por muchos textos que reunamos, habrá frases razonables que nunca hayamos encontrado exactamente así.

Piensa en una frase que acabas de inventar para describir lo que has hecho esta mañana. Es posible que nadie haya escrito nunca esa misma secuencia y, aun así, se entiende perfectamente. Necesitamos un método que aproveche las regularidades aprendidas para trabajar también con combinaciones nuevas. Ahí empieza a resultar útil la idea de sustituir una tabla por una función.

## **12. De guardar resultados a aprender una relación**

Observa los pares de números: al 0 le corresponde 0, al 1 le corresponde 1, al 2 le corresponde 4 y al 3 le corresponde 9. Si seguimos, al 4 le corresponde 16 y al 5 le corresponde 25. ¿Reconoces la relación? En todos los casos multiplicamos el número por sí mismo. Esa operación se llama elevar al cuadrado y se escribe f(x) = x². Si ahora nos preguntan por 20, podemos calcular 400 aunque ese valor no aparezca en la tabla. La función nos permite obtener resultados para nuevas entradas sin guardar antes cada pareja posible. Esa es la idea que necesitamos para dar el siguiente paso: aprender una relación que podamos aplicar más allá de los ejemplos disponibles.

Con el lenguaje no encontraremos una fórmula tan sencilla ni exacta. Las relaciones son mucho más complejas. Una red neuronal aprende una función mediante ajustes numéricos, a partir de ejemplos, y utiliza esa función para estimar continuaciones. El modelo puede aprovechar semejanzas entre situaciones. Haber aprendido cómo se construyen muchas peticiones y explicaciones le ayuda a responder a una instrucción que nunca haya visto con esas mismas palabras. A esta capacidad de aplicar lo aprendido a casos nuevos la llamamos generalización. Generalizar no garantiza acertar. Podemos reconocer un patrón donde no corresponde o aplicar una regularidad fuera de las situaciones en que funciona bien. Lo importante es que ya no dependemos exclusivamente de encontrar una casilla con una secuencia exacta: podemos calcular una estimación a partir de relaciones aprendidas.

## **13. Cómo convertir el texto en números**

<div style="text-align:center">
<img src="imgs/llm/llm17.jpg" alt="Diapositiva 20. La red neuronal utiliza representaciones numéricas para estimar la continuación" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 20. La red neuronal utiliza representaciones numéricas para estimar la continuación</em></p>

Una red neuronal realiza operaciones con números. Por tanto, antes de procesar una frase necesitamos convertirla en una representación numérica. ¿Cómo pasamos de «tengo que hacer» a algo con lo que pueda calcular una máquina?

El primer paso consiste en dividir el texto en tokens. Un token puede corresponder a una palabra completa, a una parte de una palabra o a un signo. No conviene identificar token y palabra como si fueran siempre lo mismo: una palabra larga o poco habitual puede ocupar varias unidades.

Cada token tiene un identificador. A partir de él se obtiene una lista de números que representa esa unidad. A una lista de números la llamamos vector; cuando se aprende como representación de una unidad, hablamos de embedding. Puedes imaginarlo como una forma de situar cada token en un espacio numérico donde el modelo puede realizar sus cálculos.

No necesitamos asignar un significado sencillo a cada número. Lo importante es que las representaciones permitan aprender relaciones útiles. A medida que avanzan por la red, se transforman y se combinan con información del contexto.

Volvamos a la palabra «banco». En «me senté en el banco» aparece relacionada con una acción y una situación diferentes de las de «el banco aprobó la operación». Aunque partamos del mismo token, las representaciones internas pueden incorporar esas diferencias de contexto.

La expresión y = f(x₁, x₂, …, xₙ) resume la idea general: una función recibe valores de entrada y produce un resultado. En un modelo generativo nos interesa obtener probabilidades para los tokens que podrían venir a continuación. Después utilizaremos esas probabilidades para seleccionar una continuación y seguir construyendo el texto.

## **14. Aprender mediante el entrenamiento**

<div style="text-align:center">
<img src="imgs/llm/llm18.jpg" alt="Diapositiva 21. El preentrenamiento ajusta el modelo antes de que lo utilicemos" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 21. El preentrenamiento ajusta el modelo antes de que lo utilicemos</em></p>

Ya sabemos qué queremos calcular. Ahora falta averiguar cómo aprende la red a hacerlo. Entrenar un modelo consiste en ajustar sus parámetros utilizando muchos ejemplos.

Imagina que tenemos la frase «El gato duerme en el sofá». Podemos mostrar una parte y pedir al modelo que anticipe lo siguiente. Durante el entrenamiento conocemos la continuación real, porque ya está escrita en el texto original. Eso nos permite comparar la predicción con lo que efectivamente aparece.

El modelo asigna probabilidades a distintas alternativas. Si concede muy poca probabilidad al token que aparecía en el ejemplo, el error es mayor. Mediante un procedimiento de optimización se ajustan los parámetros para mejorar sus predicciones sobre los datos de entrenamiento.

No se trata simplemente de señalar «correcto» o «incorrecto». La probabilidad asignada importa: dar un poco más de peso a la continuación observada también puede suponer una mejora. Esos pequeños ajustes, repetidos sobre muchísimos ejemplos, van modificando la función que calcula el modelo.

El propio texto proporciona la señal de aprendizaje. No hace falta que una persona escriba a mano una pregunta y una respuesta para cada fragmento. Por eso hablamos de aprendizaje autosupervisado: a partir de los datos construimos la tarea de predecir una parte utilizando otra.

Para mejorar en esa tarea, el modelo aprende regularidades de distintos tipos. Algunas tienen que ver con la gramática; otras, con relaciones entre conceptos o con la forma habitual de organizar una explicación. Anticipar una continuación puede requerir mucho más que recordar dos palabras que suelen aparecer juntas.

## **15. Lo aprendido antes de conversar**

<div style="text-align:center">
<img src="imgs/llm/llm19.jpg" alt="Diapositiva 22. Una gran colección de textos ofrece numerosos ejemplos de uso del lenguaje" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 22. Una gran colección de textos ofrece numerosos ejemplos de uso del lenguaje</em></p>

Imagina la cantidad de expresiones que encontrarías al recorrer una biblioteca: narraciones, descripciones y explicaciones sobre temas muy distintos. Cada texto ofrece ejemplos de cómo se utiliza el lenguaje. Al reunir muchos materiales, multiplicamos las oportunidades de aprender sus regularidades.

Ese aprendizaje previo explica el término «preentrenado». Cuando empezamos una conversación con un modelo, ya se ha realizado una etapa extensa de ajuste de sus parámetros. No parte de cero cada vez que abrimos una consulta.

Durante el entrenamiento, los textos sirven para modificar esos valores numéricos. Después, al generar una respuesta, el modelo utiliza los parámetros aprendidos. No necesita recorrer una biblioteca y encontrar una página exacta para cada frase que escribe.

Puede memorizar algunos fragmentos, además de aprender regularidades generales. Sin embargo, eso no le proporciona un catálogo fiable de la procedencia de todas sus respuestas. Si queremos saber de dónde sale una afirmación, necesitamos fuentes que podamos consultar y referencias que realmente la respalden.

Para desarrollar un asistente conversacional suelen añadirse otras etapas de ajuste. Los ejemplos de instrucciones y respuestas ayudan a aprender cómo atender una petición. Las señales de evaluación o preferencia pueden orientar el comportamiento hacia resultados más útiles. Así se prepara el modelo para una interacción distinta de continuar cualquier texto sin más indicaciones.

También es importante distinguir aprender durante el entrenamiento y utilizar información durante una conversación. Si le dices al modelo que adapte una explicación para alguien de doce años, esa indicación puede influir en su respuesta porque forma parte del contexto. No implica que sus parámetros se modifiquen de inmediato y de manera permanente.

La memoria que pueda gestionar una aplicación y los procesos posteriores de mejora del servicio son mecanismos diferentes. Para entender el uso ordinario, basta con recordar que estamos aplicando un modelo ya entrenado a una entrada nueva. Ese uso recibe el nombre de inferencia.

<p style="text-align:center"><em>Diapositiva 24. La petición se incorpora al contexto utilizado para generar la respuesta</em></p>

Vamos a escribir una petición muy sencilla: «Dos más dos». La aplicación prepara la información que recibirá el modelo. Además de esas palabras, puede incluir instrucciones generales y mensajes anteriores de la conversación.

Llamamos contexto al conjunto de información disponible para producir la respuesta. La frase que acabamos de escribir es una parte de ese conjunto. Por eso una petición breve puede dar lugar a resultados diferentes según lo que hayamos pedido antes.

Si estamos completando enunciados, «Dos más dos» puede continuar con «son cuatro». Si hemos pedido traducir cada expresión al inglés, la respuesta esperada cambia. El contenido de la frase es el mismo, pero la tarea que debe realizar el modelo es distinta.

Antes de generar la respuesta, el texto se convierte en tokens y se obtienen sus representaciones numéricas. A partir de ahí, el modelo estima probabilidades para los posibles tokens siguientes. Vamos a seguir el proceso utilizando palabras completas para que resulte más fácil de ver.

El contexto tiene un tamaño limitado. Un sistema puede gestionar una conversación larga seleccionando o resumiendo partes de ella, pero no conviene suponer que todo lo dicho estará siempre disponible con idéntico detalle. La respuesta depende de la información que el modelo recibe en ese momento.

Esto nos da una primera orientación práctica. Si queremos una explicación adaptada a un público concreto o basada en un documento, ayuda que esa información esté disponible y que la tarea quede clara. Cuanto mejor delimitemos qué esperamos, más fácil será orientar la continuación.

## **17. Una respuesta se construye paso a paso**

<div style="text-align:center">
<img src="imgs/llm/llm21.jpg" alt="Diapositiva 25. El punto de partida es la secuencia Dos más dos" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 25. El punto de partida es la secuencia Dos más dos</em></p>

Tenemos «Dos más dos» y todavía no se ha generado la respuesta. El modelo calcula qué tokens podrían venir a continuación y asigna una probabilidad a cada uno. Después, el sistema utiliza esas probabilidades para elegir una continuación.

<div style="text-align:center">
<img src="imgs/llm/llm22.jpg" alt="Diapositiva 26. La primera continuación es son" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 26. La primera continuación es son</em></p>

En nuestro ejemplo aparece «son». No hace falta haber escrito de antemano toda la respuesta para empezar a producirla: el texto se va construyendo paso a paso.

La elección no tiene por qué limitarse siempre a tomar la alternativa con mayor probabilidad. También se puede seleccionar entre varias opciones mediante muestreo. Esto permite distintas redacciones y explica por qué una misma petición puede recibir respuestas diferentes sin que el modelo haya aprendido algo nuevo entre ambas.

¿Qué ocurre después de generar «son»? Esa unidad se añade a lo que ya estaba escrito. Ahora la secuencia es «Dos más dos son». El modelo utiliza el contexto ampliado para calcular la siguiente continuación.

<div style="text-align:center">
<img src="imgs/llm/llm23.jpg" alt="Diapositiva 28. La nueva secuencia permite continuar con cuatro" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 28. La nueva secuencia permite continuar con cuatro</em></p>

Con «Dos más dos son» como punto de partida, el siguiente resultado de nuestro ejemplo es «cuatro». Una vez añadido, tendremos «Dos más dos son cuatro». Y el proceso podría repetirse otra vez.

Esta forma de producir texto se llama generación autorregresiva. El nombre puede resultar poco familiar, pero la idea es sencilla: cada paso utiliza las unidades anteriores, incluidas las que el propio modelo acaba de generar.

En una respuesta larga sucede lo mismo. Se va construyendo una secuencia y cada elección influye en las que vendrán después. Generar una unidad cada vez tampoco significa mirar solo la última palabra: el cálculo puede utilizar información de un contexto mucho más amplio.

Hemos elegido una suma fácil para concentrarnos en el proceso de generación. En una tarea matemática más compleja, una aplicación puede apoyarse en una calculadora o ejecutar código. Esas herramientas permiten obtener y comprobar cálculos mediante procedimientos distintos de la generación lingüística.

## **18. Cómo sabe cuándo terminar**

<div style="text-align:center">
<img src="imgs/llm/llm24.jpg" alt="Diapositiva 30. Un marcador especial indica el final de la secuencia" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 30. Un marcador especial indica el final de la secuencia</em></p>

Ya tenemos «Dos más dos son cuatro». La respuesta podría seguir, por ejemplo, con una explicación, pero también puede detenerse ahí. Para representar el final utilizamos \<eos\>, abreviatura de end of sequence, que significa «fin de secuencia».

Los sistemas concretos pueden utilizar otros marcadores de fin de turno o condiciones adicionales de parada. También pueden detenerse porque alcanzan un límite de longitud. Lo importante es que terminar forma parte del proceso: no basta con saber añadir contenido, también hay que cerrar la respuesta.

Podemos imaginar el ciclo completo: el modelo procesa el contexto, estima probabilidades para la siguiente unidad, se selecciona una, se añade a la secuencia y se vuelve a empezar. La generación continúa hasta que se cumple una condición de parada.

Ahora podemos entender la G de GPT: generativo. El modelo produce una secuencia de salida. Esa secuencia puede responder a una pregunta, desarrollar una explicación o reformular un texto que le hemos dado.

Producir contenido no garantiza que sea verdadero ni que cada frase sea completamente original. El resultado depende de lo aprendido, de la información disponible y del proceso de selección de las continuaciones. Veremos enseguida cómo influye esto en la forma de interpretar las respuestas.

## **19. Cómo se relacionan las partes del contexto**

<div style="text-align:center">
<img src="imgs/llm/llm25.jpg" alt="Diapositiva 31. Una red neuronal transforma las entradas mediante operaciones y conexiones aprendidas" width="60%" style="max-width:100%;height:auto;">
</div>

<p style="text-align:center"><em>Diapositiva 31. Una red neuronal transforma las entradas mediante operaciones y conexiones aprendidas</em></p>

Nos queda la T de GPT: transformer. Una arquitectura es una manera de organizar las operaciones y componentes de una red neuronal. La arquitectura transformer se presentó en 2017 y utiliza un mecanismo llamado atención para combinar información de distintas posiciones de una secuencia.

Pensemos en esta frase: «La profesora dejó el libro sobre la mesa porque quería consultarlo después». Para interpretar «consultarlo», nos ayuda relacionarlo con «el libro», aunque haya varias palabras entre ambos. El lenguaje está lleno de relaciones que no se limitan a la palabra inmediatamente anterior.

El mecanismo de atención permite combinar representaciones de distintas posiciones asignándoles diferentes pesos. Así, algunas contribuyen más que otras a construir una representación nueva. A través del entrenamiento, el modelo aprende a utilizar esas operaciones de una forma útil para su tarea.

La palabra «atención» se refiere aquí a un cálculo matemático. No necesitamos suponer que el sistema presta atención como una persona para comprender qué hace: establece relaciones entre representaciones y utiliza sus resultados en las siguientes operaciones.

En los modelos GPT se utiliza atención causal. Una posición puede trabajar con información de posiciones anteriores y de la propia posición, pero no con tokens posteriores de la secuencia. Durante la generación, estos todavía no se han producido. El texto nuevo se calcula a partir de lo disponible.

La red transforma la información a lo largo de varias capas. Cada una aplica operaciones que contribuyen a construir las representaciones utilizadas para predecir la continuación. La atención es una parte esencial de esa arquitectura, junto con otras transformaciones numéricas.

Ya podemos reunir las piezas: el modelo se ha preentrenado con numerosos ejemplos, utiliza una arquitectura transformer para procesar el contexto y genera una respuesta añadiendo unidades sucesivas. Las tres palabras de Generative Pretrained Transformer describen aspectos complementarios de ese funcionamiento.

## **20. Respuestas útiles y errores convincentes**

Ahora resulta más fácil comprender por qué una herramienta puede explicar bien muchas cosas y, aun así, equivocarse. El modelo ha aprendido regularidades que le permiten construir continuaciones útiles. Pero que una frase encaje en un contexto no basta para demostrar que describe un hecho verdadero.

Imagina que le pedimos el título de un artículo científico sobre un tema muy concreto. Puede producir un título verosímil, añadir nombres de autores y dar al conjunto un aspecto convincente. Si ese artículo no existe, la respuesta es incorrecta aunque tenga la forma que esperábamos. Este tipo de contenido falso o sin respaldo suele recibir el nombre de alucinación.

La información aportada durante la consulta puede ayudar. Si damos un documento y pedimos localizar una fecha, reducimos la tarea a un material concreto. También tendremos una fuente a la que volver para comprobar la respuesta. Aun así, el modelo puede pasar por alto una condición o interpretar mal un dato.

Para aprender, estas herramientas pueden ser útiles cuando necesitamos otra explicación, un ejemplo adicional o una pregunta con la que comprobar lo que hemos entendido. Podemos pedir: «Explícame la diferencia con un caso cotidiano» o «Señala qué parte de mi razonamiento no se deduce de los datos». Después conviene contrastar el resultado con el material de estudio.

Una buena petición ayuda a aclarar el objetivo. Compara «Explícame esto» con «Explica este concepto para alguien que acaba de empezar, utiliza un ejemplo sencillo y aclara primero las palabras técnicas». En el segundo caso hemos proporcionado información que orienta mejor la respuesta.

Si trabajamos con documentos, podemos pedir que distinga lo que aparece en ellos de lo que no puede concluirse. Si necesitamos un cálculo, podemos comprobarlo con una herramienta adecuada. Y si buscamos un dato factual, podemos acudir a una fuente que permita verificarlo.

La idea de fondo ya la conoces: el modelo aprende relaciones a partir de ejemplos y las aplica al contexto para generar una continuación. Esa combinación explica tanto buena parte de su utilidad como algunas de sus limitaciones. Comprenderla nos ayuda a conversar con estas herramientas sabiendo mejor qué podemos pedirles y cómo valorar lo que nos devuelven.

## **Glosario breve**

**ChatGPT.** Aplicación conversacional que permite interactuar con modelos de lenguaje y con otros componentes del sistema.

**LLM.** Siglas de Large Language Model, «gran modelo de lenguaje». Aprende regularidades a partir de datos y las utiliza para procesar y generar lenguaje.

**Token.** Unidad en la que se divide el texto para procesarlo. Puede ser una palabra, parte de una palabra o un signo, entre otras posibilidades.

**Corpus.** Colección de textos que se utiliza para estudiar el lenguaje o desarrollar un modelo.

**Parámetro.** Valor numérico que interviene en los cálculos del modelo y se ajusta durante el entrenamiento.

**Vector y embedding.** Un vector es una lista de números. Un embedding es una representación numérica aprendida de una unidad, como un token.

**Contexto.** Información que recibe el modelo para producir una respuesta. Puede incluir instrucciones, mensajes anteriores y documentos.

**Preentrenamiento.** Etapa previa de aprendizaje a partir de grandes cantidades de datos, antes de otros ajustes y del uso ordinario del modelo.

**Inferencia.** Uso de un modelo ya entrenado para obtener un resultado a partir de una entrada nueva.

**Generación autorregresiva.** Producción de una secuencia paso a paso, utilizando las unidades anteriores para generar cada nueva unidad.

**Transformer.** Arquitectura neuronal que combina mecanismos de atención con otras operaciones para procesar las representaciones de una secuencia.

**RAG.** Generación aumentada con recuperación. Consiste en recuperar información pertinente e incorporarla al contexto con el que se genera una respuesta.
